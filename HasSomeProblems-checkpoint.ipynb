```python
import pandas as pd
from tensorflow.keras.preprocessing.text import Tokenizer

df = pd.read_csv("data/augmented_df.csv")
df.head()
```

```python
import re

def clean_text(t):
    t = t.lower()
    t = re.sub(r"[^a-z0-9\s]", " ", t)
    t = re.sub(r"\s+", " ", t).strip()
    return t

df["symptoms"] = df["symptoms"].apply(clean_text)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y = le.fit_transform(df["condition"])

print("Number of classes:", len(le.classes_))
```

```python
import joblib
joblib.dump(le, "models/label_encoder.pkl")
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

MAX_WORDS = 8000

tokenizer = Tokenizer(num_words=MAX_WORDS, oov_token="<OOV>")
tokenizer.fit_on_texts(df["symptoms"])

sequences = tokenizer.texts_to_sequences(df["symptoms"])
X_lstm = pad_sequences(sequences, maxlen=120, padding="post")

print("Tokenizer ready! Vocab size:", len(tokenizer.word_index))
```

```python
import joblib
joblib.dump(tokenizer, "models/lstm_tokenizer.pkl")
```

```python
import numpy as np

np.save("data/features/X_lstm.npy", X_lstm)
```

```python
# PART2 distilbert tokenizer

from transformers import DistilBertTokenizerFast, DistilBertModel
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

bert_tokenizer = DistilBertTokenizerFast.from_pretrained("distilbert-base-uncased")
bert_model = DistilBertModel.from_pretrained("distilbert-base-uncased").to(device)
bert_model.eval()
```

```python
import numpy as np

batch_size = 32
bert_embeddings = []

texts = df["symptoms"].tolist()

for i in range(0, len(texts), batch_size):
    batch = texts[i:i+batch_size]
    enc = bert_tokenizer(batch, padding=True, truncation=True, max_length=128, return_tensors="pt")
    
    input_ids = enc["input_ids"].to(device)
    attention_mask = enc["attention_mask"].to(device)
    
    with torch.no_grad():
        outputs = bert_model(input_ids, attention_mask=attention_mask)
        cls_vectors = outputs.last_hidden_state[:,0,:].cpu().numpy()
        bert_embeddings.append(cls_vectors)

bert_embeddings = np.vstack(bert_embeddings)
bert_embeddings.shape
```

```python
np.save("data/features/X_bert.npy", bert_embeddings)
```

```python
# PART 3 — Metadata Encoding (age/gender)

meta = np.zeros((len(df), 2))  # 2 = placeholder for "age" + "gender"
np.save("data/features/X_meta.npy", meta)
```
